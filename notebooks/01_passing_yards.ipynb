# NFL Passing Yards

How many passing yards did a QB average over 5 games?

This example uses fictional data.

```python
passing_yards = [245, 310, 198, 275, 322]
passing_yards
```

```python
total_yards = sum(passing_yards)
games_played = len(passing_yards)
average_yards = total_yards / games_played

average_yards
```

```python
import matplotlib.pyplot as plt

games = [1, 2, 3, 4, 5]
plt.bar(games, passing_yards)
plt.xlabel("Game")
plt.ylabel("Passing Yards")
plt.title("QB Passing Yards per Game")
plt.show()
```

```python
yardage_range = max(passing_yards) - min(passing_yards)
yardage_range
```

```python
import statistics

qb_a = passing_yards
qb_b = [272, 276, 270, 278, 274]

print(statistics.pstdev(qb_a))
print(statistics.pstdev(qb_b))
```

`statistics.pstdev(list)` is used when the list is the entire population

`statistics.stdev(list)` is used when the list is a sample that represents a population

```python
print(statistics.pstdev(qb_a))
print(statistics.stdev(qb_a))
```

```python
mean_yards = statistics.mean(qb_a)
deviations = [yards - mean_yards for yards in qb_a]

deviations
```

```python
# The sum of deviations should always equal zero.
sum(deviations)
```

```python
squared_deviations = [difference**2 for difference in deviations]
squared_deviations
```

Variance: lbs²

Standard Deviation: √(lbs²) = lbs

```python
variance = sum(squared_deviations) / len(squared_deviations)
standard_deviation = variance**0.5

print(variance)
print(standard_deviation)
```

```python
plt.plot(games, qb_a, label="QB A", marker="o")
plt.plot(games, qb_b, label="QB B", marker="o")
plt.xlabel("Game")
plt.ylabel("Passing Yards")
plt.title("Passing Yards Across Five Games")
plt.legend()
plt.show()
```
